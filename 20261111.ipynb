```python
def x(a,b,c):
  if b=='+':
     return a+c
  elif b=='-':
     return a-c
  elif b=='*':
     return a*c
  elif b=='/':
    if c==0:
      return 'error'
    return a/c
  else:
    return 'error'

x(1, '//', 21)
```
